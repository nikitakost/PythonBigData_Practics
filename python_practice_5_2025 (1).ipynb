```python
class ThreeDFigure:
    def get_volume(self):
        return 0  

    def __add__(self, other):
        if isinstance(other, ThreeDFigure):  
            return self.get_volume() + other.get_volume()
```

8. (13б). Створіть клас EWallet з приватними атрибутами balance та id. Реалізуйте методи поповнення та зняття коштів, забезпечивши належну інкапсуляцію. Підказка: використовуйте декоратори property та setter.

```python
class EWallet:
    def __init__(self, balance, wallet_id):
        self.__balance = balance 
        self.__id = wallet_id 

    @property
    def balance(self):
        return self.__balance

    @property
    def id(self):
        return self.__id

    def topup(self, amount):
        if amount > 0:
            self.__balance += amount
            print(f"Top up {amount} UAH. New balance: {self.__balance} UAH.")
        else:
            print("Amount must be greater than 0.")

    def withdraw(self, amount):
        if self.__balance >= amount:
            self.__balance -= amount
            print(f"Withdraw {amount} UAH. New balance: {self.__balance} UAH.")
        else:
            print("Not enough money.")


wallet = EWallet(100, 1)
wallet.topup(50)
wallet.withdraw(30)
wallet.withdraw(200)
print("Final balance:", wallet.balance)


```

9. (11б). Створіть клас ShoppingCart, який містить список об'єктів типу Item. Реалізуйте методи для додавання айтемів, їх видалення та відображення списку айтемів.

```python
class ShoppingCart:
    def __init__(self):
        self.items = []  

    def add_item(self, name, price):
        self.items.append({"name": name, "price": price})
        print(f"Added {name} to the cart.")

    def remove_item(self, name):
        for item in self.items:
            if item["name"] == name:
                self.items.remove(item)
                print(f"Removed {name} from the cart.")
                return
        print(f"Item '{name}' not found in the cart.")

    def display_items(self):
        if not self.items:
            print("The shopping cart is empty.")
        else:
            print("Shopping Cart:")
            for item in self.items:
                print(f"- {item['name']}: {item['price']:.2f}")


# Example usage:
cart = ShoppingCart()
cart.add_item("Laptop", 999.99)
cart.add_item("Headphones", 199.99)

cart.display_items()

cart.remove_item("Laptop")
cart.display_items()
```

13. (11б). Створіть клас, що має визначений Iterator Protocol та при кожному виклику next повертає назву дня тижня, поки вони не вичерпаються. Зверніть увагу, що порядок має значення.

```python
day_of_week_list = ["Monday", "Tuesday", "Wednesday", "Thursday", "Friday", "Saturday", "Sunday"]
day_of_week_iter = iter(day_of_week_list)

try:
    while True:
        print(next(day_of_week_iter))
except StopIteration:
    print("That's it")
```
